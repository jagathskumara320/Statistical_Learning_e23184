#Q3) Transformer Failures

In a certain electricity distribution district, the number of emergency transformer failures reported in a single day is modeled as a discrete random variable $X$. Assume that $X$ takes values in the set $\{0,1,2,3\}$, with probabilities

$$
P(X=0)=0.50,\qquad P(X=1)=0.30,\qquad P(X=2)=0.15,\qquad P(X=3)=0.05.
$$

1. Construct a probability space $(\Omega,\mathfrak F,P)$ suitable for this model.
2. Define $X$ as a random variable on this space.
3. Determine the probability mass function of $X$.
4. Compute the expectation $E[X]$.
5. Compute the variance $\mathrm{Var}(X)$.
6. Find $P(X\ge 1)$ and $P(X\ge 2)$.
7. Briefly explain the practical meaning of the expectation and variance in this setting.



```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------------------------------------
# Q3: Transformer Failures
# -----------------------------------------------------------

# 1. Probability space
# Omega = {0, 1, 2, 3}  (number of failures in a day)
# F     = Power set of Omega (all 16 subsets)
# P     = probability measure defined by the PMF below

# 2 & 3. Random variable X and its PMF
values = np.array([0, 1, 2, 3])
pmf    = np.array([0.50, 0.30, 0.15, 0.05])

print("=== Q3: Transformer Failures ===")
print("\nProbability Mass Function (PMF):")
for k, p in zip(values, pmf):
    print(f"  P(X = {k}) = {p}")

# 4. Expectation  E[X] = sum k * P(X=k)
E_X = np.sum(values * pmf)
print(f"\n4. E[X] = {E_X}")

# 5. Variance  Var(X) = E[X^2] - (E[X])^2
E_X2  = np.sum(values**2 * pmf)
Var_X = E_X2 - E_X**2
print(f"5. Var(X) = E[X^2] - (E[X])^2 = {E_X2} - {E_X**2:.4f} = {Var_X:.4f}")

# 6. Tail probabilities
P_ge1 = 1 - pmf[0]                  # P(X >= 1) = 1 - P(X=0)
P_ge2 = pmf[2] + pmf[3]             # P(X >= 2) = P(X=2) + P(X=3)
print(f"6. P(X >= 1) = {P_ge1}")
print(f"   P(X >= 2) = {P_ge2}")

# Visualisation
fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(values, pmf, color='steelblue', edgecolor='black', width=0.5)
ax.set_xlabel('Number of failures (X)', fontsize=12)
ax.set_ylabel('Probability', fontsize=12)
ax.set_title('Q3 – PMF of Daily Transformer Failures', fontsize=13)
ax.set_xticks(values)
for k, p in zip(values, pmf):
    ax.text(k, p + 0.01, f'{p}', ha='center', fontsize=11)
plt.tight_layout()
plt.show()

print("\n7. Interpretation:")
print(f"   E[X] = {E_X}: On average, {E_X} transformer failure(s) are expected per day.")
print(f"   Var(X) = {Var_X:.4f}: The day-to-day count fluctuates with variance {Var_X:.4f}")
print(f"   (std dev \u2248 {np.sqrt(Var_X):.4f} failures), indicating moderate variability around the mean.")
```

### Q3 – Written Answers

**1. Probability Space**  
$$\Omega = \{0,1,2,3\}, \quad \mathfrak{F} = \mathcal{P}(\Omega) \text{ (all 16 subsets)}, \quad P \text{ defined by the PMF below.}$$

**2. Random Variable**  
$X:\Omega\to\{0,1,2,3\}$ is the identity map $X(\omega)=\omega$, counting the number of failures.

**3. Probability Mass Function**  
| $k$ | $P(X=k)$ |
|-----|----------|
| 0   | 0.50     |
| 1   | 0.30     |

| 3   | 0.05     |

All probabilities are non-negative and sum to $0.50+0.30+0.15+0.05=1$. ✓

**4. Expectation**  
$$E[X]=\sum_{k=0}^{3}k\,P(X=k)=0(0.50)+1(0.30)+2(0.15)+3(0.05)=0+0.30+0.30+0.15=\boxed{0.75}$$

**5. Variance**  
$$E[X^2]=0^2(0.50)+1^2(0.30)+2^2(0.15)+3^2(0.05)=0+0.30+0.60+0.45=1.35$$
$$\mathrm{Var}(X)=E[X^2]-(E[X])^2=1.35-0.75^2=1.35-0.5625=\boxed{0.7875}$$

**6. Tail Probabilities**  
$$P(X\ge 1)=1-P(X=0)=1-0.50=\boxed{0.50}$$
$$P(X\ge 2)=P(X=2)+P(X=3)=0.15+0.05=\boxed{0.20}$$

**7. Practical Interpretation**

Days with zero failures are most likely (50%), but there is a 1-in-20 chance of three failures, so maintenance teams cannot rely solely on the average for resource planning.

# Q4) Probability of Defective Items

In a manufacturing process, the proportion of defective items in a large batch produced on a given day is modeled by a continuous random variable $X$. Since $X$ represents a proportion, it takes values in the interval $[0,1]$. Suppose that $X$ has a Beta distribution with parameters $\alpha=2$ and $\beta=18$, so that its density is

$$
f(x)=
\begin{cases}
\dfrac{1}{B(2,18)}x^{1}(1-x)^{17}, & 0<x<1,\\\
0, & \text{otherwise},
\end{cases}
$$

where $B(2,18)$ is the Beta function.

Answer the following:

1. Construct a probability space $(\Omega,\mathfrak F,P)$ suitable for this model.

2. Define $X$ as a continuous random variable on this space.

3. Write down the probability density function of $X$.

4. Verify that $f$ is a valid probability density function.

5. Compute the expectation $E[X]$.

6. Compute the variance $\mathrm{Var}(X)$.

7. Find the probability that the defective proportion is at most $0.10$, that is, compute

$$
P(X\le 0.10).
$$

You may leave your answer in Beta integral form.

8. Find the probability that the defective proportion lies between $0.05$ and $0.15$, that is, compute

$$
P(0.05\le X\le 0.15).
$$

You may leave your answer in integral form.

9. Briefly explain why a Beta distribution is a reasonable model for this situation.

10. Interpret the expectation and variance in the context of production quality.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import beta as beta_dist
from scipy.integrate import quad
from scipy.special import beta as beta_fn

# -----------------------------------------------------------
# Q4: Defective Items – Beta(2, 18) distribution
# -----------------------------------------------------------

alpha_param, beta_param = 2, 18
rv = beta_dist(alpha_param, beta_param)

print("=== Q4: Probability of Defective Items ===")

# 4. Verify f is a valid PDF (integrates to 1)
B_val = beta_fn(alpha_param, beta_param)
def f(x):
    return x**(alpha_param - 1) * (1 - x)**(beta_param - 1) / B_val

integral_val, _ = quad(f, 0, 1)
print(f"\n4. Integral of f over [0,1] = {integral_val:.6f}  (should be 1.0) \u2713")

# 5. Expectation  E[X] = alpha / (alpha + beta)
E_X = alpha_param / (alpha_param + beta_param)
print(f"\n5. E[X] = alpha/(alpha+beta) = {alpha_param}/{alpha_param+beta_param} = {E_X:.6f}")

# 6. Variance  Var(X) = alpha*beta / [(alpha+beta)^2 * (alpha+beta+1)]
n = alpha_param + beta_param
Var_X = (alpha_param * beta_param) / (n**2 * (n + 1))
print(f"6. Var(X) = {alpha_param}*{beta_param}/({n}^2*{n+1}) = {Var_X:.8f}")

# 7. P(X <= 0.10)
P_le_010 = rv.cdf(0.10)
print(f"\n7. P(X <= 0.10) = {P_le_010:.6f}")

# 8. P(0.05 <= X <= 0.15)
P_005_015 = rv.cdf(0.15) - rv.cdf(0.05)
print(f"8. P(0.05 <= X <= 0.15) = {P_005_015:.6f}")

# Plot the PDF
x_vals = np.linspace(0, 0.5, 500)
y_vals = rv.pdf(x_vals)

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(x_vals, y_vals, color='darkorange', lw=2, label=r'Beta(2, 18) PDF')

# Shade P(X <= 0.10)
x_shade1 = np.linspace(0, 0.10, 300)
ax.fill_between(x_shade1, rv.pdf(x_shade1), alpha=0.3, color='steelblue',
                label=f'P(X≤0.10) = {P_le_010:.4f}')

# Shade P(0.05 <= X <= 0.15)
x_shade2 = np.linspace(0.05, 0.15, 300)
ax.fill_between(x_shade2, rv.pdf(x_shade2), alpha=0.4, color='green',
                label=f'P(0.05≤X≤0.15) = {P_005_015:.4f}')

ax.axvline(E_X, color='red', linestyle='--', label=f'E[X] = {E_X:.3f}')
ax.set_xlabel('Defective proportion (X)', fontsize=12)
ax.set_ylabel('Density f(x)', fontsize=12)
ax.set_title('Q4 – Beta(2, 18) PDF: Defective Proportion', fontsize=13)
ax.legend(fontsize=10)
plt.tight_layout()
plt.show()
```

### Q4 – Written Answers

**1. Probability Space**  
$$\Omega=(0,1),\quad \mathfrak{F}=\mathcal{B}((0,1))\text{ (Borel }σ\text{-algebra on }(0,1)),\quad P\text{ induced by the Beta}(2,18)\text{ density.}$$

**2. Random Variable**  
$X:\Omega\to(0,1)$ is the identity map $X(\omega)=\omega$, representing the daily defective proportion.

**3. Probability Density Function**  
$$f(x)=\frac{x^{1}(1-x)^{17}}{B(2,18)},\quad 0<x<1,\quad 0\text{ otherwise.}$$

**4. Validity of PDF**  
By definition of the Beta function:
$$\int_0^1 f(x)\,dx = \frac{1}{B(2,18)}\int_0^1 x^{1}(1-x)^{17}\,dx = \frac{B(2,18)}{B(2,18)}=1.$$
Numerical verification gives $\int_0^1 f(x)\,dx = 1.000000$ \checkmark

**5. Expectation**  
For Beta$(\alpha,\beta)$: $E[X]=\dfrac{\alpha}{\alpha+\beta}=\dfrac{2}{20}=\boxed{0.10}$

**6. Variance**  
$$\mathrm{Var}(X)=\frac{\alpha\beta}{(\alpha+\beta)^2(\alpha+\beta+1)}=\frac{2\cdot 18}{20^2\cdot 21}=\frac{36}{8400}\approx\boxed{0.004286}$$

**7. $P(X\le 0.10)$**  
$$P(X\le 0.10)=\frac{1}{B(2,18)}\int_0^{0.10}x(1-x)^{17}\,dx\approx\boxed{0.5797}$$

**8. $P(0.05\le X\le 0.15)$**  
$$P(0.05\le X\le 0.15)=\frac{1}{B(2,18)}\int_{0.05}^{0.15}x(1-x)^{17}\,dx\approx\boxed{0.5562}$$

**9. Why Beta is Reasonable**  
The Beta distribution is defined on $[0,1]$, making it a natural model for any proportion. With $\alpha=2,\beta=18$, it is right-skewed, assigning most probability to small defective proportions (close to 0), which reflects the expectation that a manufacturing process mostly produces good items. The two shape parameters can be tuned to historical data, offering great flexibility.

**10. Practical Interpretation**  
- **$E[X]=0.10$**: The expected daily defective proportion is 10%, i.e., 1 in 10 items is defective on average.  
- **$\mathrm{Var}(X)\approx 0.0043$** (std dev $\approx 0.0655$): Day-to-day variation in the defective rate is small, so the process is relatively stable around 10%. Quality managers can set control limits based on this spread.

# Q5) Expected Project Completion Time

A construction company is estimating the time required to install a backup generator at a hospital. Let $X$ denote the number of hours needed to complete the installation. Based on expert judgment, the shortest possible time is $4$ hours, the longest possible time is $10$ hours, and the most likely completion time is $6$ hours. Assume that $X$ follows a Triangular distribution with parameters $a=4$, $b=10$, and $c=6$.

Answer the following:

#Q6) Expected Project Completion Time Using PERT

A project manager is estimating the time required to complete a critical software integration task. Let $X$ denote the number of days required to complete the task. Based on expert judgment, the shortest possible time is $a=8$ days, the longest possible time is $b=20$ days, and the most likely completion time is $c=11$ days. Assume that $X$ follows a PERT distribution with parameters $(a,b,c)$.

The PERT distribution is defined as a Beta distribution scaled to the interval $[a,b]$, with internal shape parameters

$$
\alpha = 1 + 4\left( \frac{c-a}{b-a} \right)
$$

and

$$
\beta = 1 + 4\left( \frac{b-c}{b-a} \right).
$$

Its probability density function is

$$
f(x)=\frac{(x-a)^{\alpha-1}(b-x)^{\beta-1}}{B(\alpha,\beta)(b-a)^{\alpha+\beta-1}}, \qquad a \leq x \leq b,
$$

where

$$
B(\alpha,\beta)=\int_0^1 t^{\alpha-1}(1-t)^{\beta-1}\,dt.
$$

Answer the following:

1. Compute the values of the shape parameters $\alpha$ and $\beta$.

2. Specify the support of $X$.

3. Write down the probability density function $f_X(x)$ for this model.

4. State the probability space $(\Omega,\mathfrak F,P)$ suitable for this continuous model.

5. Define $X$ as a random variable on this space.

6. Show that the expectation $E[X]$ is given by

$$
E[X]=\frac{a+4c+b}{6}.
$$

7. Compute the variance $\mathrm{Var}(X)$ using the approximation

$$
\mathrm{Var}(X)=\frac{(E[X]-a)(b-E[X])}{7}.
$$

8. Compute the approximate standard deviation $\sigma$.

9. Briefly explain why a PERT distribution may be more appropriate than a Triangular distribution in this setting byt comparing the two results.

10. Interpret the practical meaning of the values $a$, $b$, $c$, $E[X]$, and $\sigma$ in the context of project planning.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import beta as beta_dist
from scipy.stats import triang
from scipy.special import beta as beta_fn

# -----------------------------------------------------------
# Q6: PERT Distribution  a=8, b=20, c=11
# -----------------------------------------------------------

a, b, c = 8, 20, 11

print("=== Q6: PERT Distribution ===")
print(f"Parameters: a = {a}, b = {b}, c = {c}")

# 1. Shape parameters
alpha = 1 + 4 * (c - a) / (b - a)
beta_p  = 1 + 4 * (b - c) / (b - a)
print(f"\n1. alpha = 1 + 4*({c}-{a})/({b}-{a}) = {alpha:.4f}")
print(f"   beta  = 1 + 4*({b}-{c})/({b}-{a}) = {beta_p:.4f}")

# 2. Support
print(f"\n2. Support of X: [{a}, {b}]")

# 3 & 4. PDF written below in markdown; scipy setup
# Scale Beta(alpha, beta) to [a, b]
rv_pert = beta_dist(alpha, beta_p, loc=a, scale=b - a)

# 6. Expectation
E_X = (a + 4*c + b) / 6
print(f"\n6. E[X] = (a + 4c + b)/6 = ({a} + 4*{c} + {b})/6 = {E_X:.4f} days")

# Derivation check via Beta mean formula: E = a + (b-a)*alpha/(alpha+beta)
E_check = a + (b - a) * alpha / (alpha + beta_p)
print(f"   Verification via Beta mean: a + (b-a)*alpha/(alpha+beta) = {E_check:.4f} \u2713")

# 7. Variance (PERT approximation)
Var_X = (E_X - a) * (b - E_X) / 7
print(f"\n7. Var(X) = (E[X]-a)(b-E[X])/7 = ({E_X:.4f}-{a})({b}-{E_X:.4f})/7 = {Var_X:.4f}")

# 8. Standard deviation
sigma = np.sqrt(Var_X)
print(f"8. sigma = sqrt(Var(X)) = {sigma:.4f} days")

# Compare with Triangular
E_tri = (a + b + c) / 3
Var_tri = (a**2 + b**2 + c**2 - a*b - a*c - b*c) / 18
print(f"\n   -- Triangular comparison --")
print(f"   E[X]_tri = {E_tri:.4f} days, Var_tri = {Var_tri:.4f}, sigma_tri = {np.sqrt(Var_tri):.4f}")
print(f"   E[X]_PERT= {E_X:.4f} days, Var_PERT= {Var_X:.4f}, sigma_PERT= {sigma:.4f}")

# Plot PERT vs Triangular
x_vals = np.linspace(a, b, 500)
c_norm_tri = (c - a) / (b - a)
rv_tri = triang(c_norm_tri, loc=a, scale=b - a)

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(x_vals, rv_pert.pdf(x_vals), color='darkorange', lw=2, label='PERT PDF')
ax.plot(x_vals, rv_tri.pdf(x_vals), color='steelblue', lw=2, linestyle='--', label='Triangular PDF')
ax.axvline(E_X,  color='darkorange', linestyle=':', label=f'E[X]_PERT = {E_X:.2f}')
ax.axvline(E_tri, color='steelblue',  linestyle=':', label=f'E[X]_Tri  = {E_tri:.2f}')
ax.set_xlabel('Completion time (days)', fontsize=12)
ax.set_ylabel('Density', fontsize=12)
ax.set_title('Q6 – PERT vs Triangular Distribution', fontsize=13)
ax.legend(fontsize=10)
plt.tight_layout()
plt.show()
```

### Q6 – Written Answers

**1. Shape Parameters**
$$\alpha = 1+4\cdot\frac{c-a}{b-a}=1+4\cdot\frac{11-8}{20-8}=1+1=\boxed{2}$$
$$\beta = 1+4\cdot\frac{b-c}{b-a}=1+4\cdot\frac{20-11}{20-8}=1+3=\boxed{4}$$

**2. Support**  
$$\text{Support of }X=[a,b]=[8,20].$$

**3. PDF**  
$$f_X(x)=\frac{(x-8)^{1}(20-x)^{3}}{B(2,4)\cdot 12^{5}},\quad 8\le x\le 20.$$

**4. Probability Space**  
$$\Omega=[8,20],\quad \mathfrak{F}=\mathcal{B}([8,20]),\quad P\text{ induced by the PERT density }f_X.$$

**5. Random Variable**  
$X:\Omega\to[8,20]$ is the identity map, representing the number of days to complete the task.

**6. Derivation of $E[X]$**  
The PERT variable is a rescaled Beta. For $Z\sim\text{Beta}(\alpha,\beta)$ we have $E[Z]=\frac{\alpha}{\alpha+\beta}$. Since $X=a+(b-a)Z$:
$$E[X]=a+(b-a)\cdot\frac{\alpha}{\alpha+\beta}.$$
Substituting $\alpha=1+4\frac{c-a}{b-a}$ and $\beta=1+4\frac{b-c}{b-a}$, so $\alpha+\beta=6$:
$$E[X]=a+(b-a)\cdot\frac{1+4\frac{c-a}{b-a}}{6}=a+\frac{(b-a)+4(c-a)}{6}=\frac{6a+(b-a)+4(c-a)}{6}=\frac{a+4c+b}{6}.$$
Numerically: $E[X]=\frac{8+44+20}{6}=\frac{72}{6}=\boxed{12\text{ days}}$.

**7. Variance**  
$$\mathrm{Var}(X)=\frac{(12-8)(20-12)}{7}=\frac{4\cdot 8}{7}=\frac{32}{7}\approx\boxed{4.571}$$

**8. Standard Deviation**  
$$\sigma=\sqrt{4.571}\approx\boxed{2.138\text{ days}}$$

**9. PERT vs. Triangular**  
For the same $(a,b,c)$, the Triangular distribution gives $E[X]=(8+20+11)/3=13$ days and $\sigma\approx 2.49$ days, while the PERT gives $E[X]=12$ days and $\sigma\approx 2.14$ days. The PERT weights the mode four times more heavily in the mean formula, pulling the expected value closer to the most-likely time. Its bell-shaped, smooth density is also more realistic than the Triangular’s sharp peak at the mode; real task durations rarely change direction instantaneously.

**10. Practical Interpretation**  
- $a=8$: Optimistic (fastest possible) completion, assuming everything goes perfectly.  
- $b=20$: Pessimistic (slowest) completion under worst-case conditions.  
- $c=11$: Most likely scenario under normal working conditions.  
- $E[X]=12$ days: The planned schedule baseline; management should allocate at least 12 days.  
- $\sigma\approx 2.14$ days: A one-standard-deviation buffer (12 + 2.14 $\approx$ 14.1 days) covers most realistic delays without over-padding the schedule.

#Q7) If the PERT Expectation is to Match Our Intuition

A project manager models the duration $X$ of a task by a PERT distribution with parameters $a$, $b$, and $c$, where $a$ is the minimum time, $b$ is the maximum time, and $c$ is the most likely time.

Suppose the minimum time is $a=5$ days and the most likely time is $c=11$ days. If the expected time computed from the PERT model is also to be the same as the most likely time:

1. Determine the value of the maximum time $b$.

2. What happens to the variance?

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import beta as beta_dist
from sympy import symbols, solve, simplify, sqrt, Rational

# -----------------------------------------------------------
# Q7: PERT with E[X] = c  =>  find b, analyse variance
# -----------------------------------------------------------

print("=== Q7: PERT Expectation Equals the Mode ===")

a_val, c_val = 5, 11

# PERT expectation formula: E[X] = (a + 4c + b) / 6 = c
# => a + 4c + b = 6c
# => b = 6c - a - 4c = 2c - a

b_val = 2*c_val - a_val
E_val = (a_val + 4*c_val + b_val) / 6

print(f"\nGiven: a = {a_val}, c = {c_val}")
print(f"Condition: E[X] = c  =>  (a + 4c + b)/6 = c")
print(f"           a + 4c + b = 6c")
print(f"           b = 2c - a = 2*{c_val} - {a_val} = {b_val}")
print(f"\n1. b = {b_val} days")
print(f"   Verification: E[X] = ({a_val} + 4*{c_val} + {b_val})/6 = {E_val:.4f} = c \u2713")

# Variance with this b
Var_val = (E_val - a_val) * (b_val - E_val) / 7
sigma_val = np.sqrt(Var_val)
print(f"\n2. Var(X) = (E[X] - a)(b - E[X]) / 7")
print(f"         = ({E_val} - {a_val})({b_val} - {E_val}) / 7")
print(f"         = {E_val - a_val} * {b_val - E_val} / 7")
print(f"         = {Var_val:.4f}")
print(f"   sigma = {sigma_val:.4f} days")

# General symbolic analysis: show Var = (c-a)^2 / 7
print(f"\n   General formula when E[X]=c and b=2c-a:")
print(f"   Var(X) = (c-a)(b-c)/7 = (c-a)(2c-a-c)/7 = (c-a)^2/7")
print(f"   = ({c_val}-{a_val})^2/7 = {(c_val-a_val)**2}/7 = {(c_val-a_val)**2/7:.4f} \u2713")

# Visualisation: symmetric PERT when E[X]=c
alpha_q7 = 1 + 4*(c_val - a_val)/(b_val - a_val)
beta_q7  = 1 + 4*(b_val - c_val)/(b_val - a_val)
rv_q7 = beta_dist(alpha_q7, beta_q7, loc=a_val, scale=b_val - a_val)

x_vals = np.linspace(a_val - 0.2, b_val + 0.2, 500)
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(x_vals, rv_q7.pdf(x_vals), color='teal', lw=2.5, label=f'PERT(a={a_val}, b={b_val}, c={c_val})')
ax.axvline(c_val, color='red', linestyle='--', lw=1.5, label=f'E[X] = mode = c = {c_val}')
ax.fill_between(x_vals, rv_q7.pdf(x_vals), alpha=0.15, color='teal')
ax.set_xlabel('Duration (days)', fontsize=12)
ax.set_ylabel('Density', fontsize=12)
ax.set_title(f'Q7 \u2013 PERT Distribution when E[X] = c\n(a={a_val}, b={b_val}, c={c_val})', fontsize=13)
ax.legend(fontsize=11)
plt.tight_layout()
plt.show()
```

### Q7 – Written Answers

**Setup**  
The PERT expectation formula is:
$$E[X]=\frac{a+4c+b}{6}.$$
We require $E[X]=c$ with $a=5$, $c=11$.

**1. Finding $b$**  
$$\frac{a+4c+b}{6}=c \implies a+4c+b=6c \implies b=2c-a.$$
$$b=2(11)-5=\boxed{17\text{ days}}.$$

*Verification:* $E[X]=(5+44+17)/6=66/6=11=c$ \checkmark

**Geometry of the solution**: $b=2c-a$ means the mode $c$ is exactly the midpoint of $[a,b]$ (since $(a+b)/2=(5+17)/2=11=c$). This makes the scaled Beta distribution symmetric around $c$, so the mean equals the mode.

**2. What Happens to the Variance?**  
Using the PERT variance approximation with $E[X]=c$ and $b=2c-a$:
$$\mathrm{Var}(X)=\frac{(E[X]-a)(b-E[X])}{7}=\frac{(c-a)(b-c)}{7}=\frac{(c-a)(2c-a-c)}{7}=\frac{(c-a)^2}{7}.$$

Numerically:
$$\mathrm{Var}(X)=\frac{(11-5)^2}{7}=\frac{36}{7}\approx\boxed{5.143},\quad \sigma\approx 2.268\text{ days}.$$

**Interpretation**: The variance is completely determined by how far the mode is from the minimum ($c-a=6$ days). Requiring $E[X]=c$ fixes $b$ symmetrically, and the spread of the distribution is entirely governed by the gap $(c-a)$. A larger gap (more optimistic mode, or lower minimum) inflates the variance; a smaller gap reduces it. There is no additional freedom to adjust the variance independently once we impose $E[X]=c$.
